```python
import requests
import pandas as pd

url = "https://voksenuddannelse.dk/soeg-api/api/search/hold/searchHold"

params = {
    "subject_code": "22906",
    "level": "-",
    "type": "AMU",
    "pageCount": "1"
}

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "application/json"
}

response = requests.get(
    url,
    params=params,
    headers=headers,
    timeout=60
)

response.raise_for_status()

data = response.json()

rows = data["holdCardDtos"]

df = pd.DataFrame(rows)

df["ledigePladser"] = (
    df["participantCapacity"] -
    df["currentParticipantAmount"]
)

print(
    df[
        [
            "holdTitle",
            "currentParticipantAmount",
            "participantCapacity",
            "ledigePladser"
        ]
    ]
    .sort_values("ledigePladser")
    .head(50)
)

print(df.columns.tolist())

print("\nFørste række:")
print(df.iloc[0])

print("\nKOLONNER:")
for col in df.columns:
    print(col)
```
